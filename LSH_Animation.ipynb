```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

from IPython.display import HTML
from matplotlib import animation

plt.rcParams['text.usetex'] = True
plt.rcParams['text.latex.preamble'] = r'\usepackage{amsfonts}'
plt.rcParams['figure.dpi'] = 200

import warnings
warnings.filterwarnings('ignore')

rng = np.random.default_rng(seed=12)
```

```python
plt.rcParams['font.size'] = 8
f_w = 2.2
artists = []
frame = []

a = np.array([1, 1])

fig, ax = plt.subplots(figsize=(3, 3))

frame.append(ax.axhline(0, color='black', linewidth=0.2))
frame.append(ax.axvline(0, color='black', linewidth=0.2))

ax.set_aspect('equal')
ax.set_xlim(-f_w, f_w)
ax.set_ylim(-f_w, f_w)

artists.append(list(frame))

frame.append(ax.arrow(0, 0, a[0], a[1], length_includes_head=True, head_width=0.05, color='black'))

artists.append(list(frame))

a = np.array([1, 1])
b = 0

p = np.array([-a[1], a[0]]) * 10 / a.min()
lx, ly = np.array([p[0], -1*p[0]]), np.array([p[1], -1*p[1]])

for c in range(2):
    frame += ax.plot(c*a[0]/(a@a)+lx,c*a[1]/(a@a)+ly+b, linewidth=0.8, label=f'$a^Tx={c}$')
    artists.append(list(frame)+[ax.legend()])

frame.append(ax.scatter([a[0]/(a@a)], [a[1]/(a@a)], marker='x', color='red'))
frame.append(ax.text(a[0]/(a@a)+0.25, a[1]/(a@a)-0.1, r'$\frac{1}{a^Ta} a$'))

artists.append(frame)

plt.close()
anim = animation.ArtistAnimation(fig, artists, interval=600, repeat=False, blit=False);
HTML(anim.to_jshtml())
```

```python
# save animation frames as pngs
anim.save('hyperplane/frame.png', writer='imagemagick')
```

```python
plt.rcParams['font.size'] = 8
f_w = 2.2
artists = []
frame = []

fig, ax = plt.subplot_mosaic("AACCCC;BBCCCC;DEFGHI", gridspec_kw={'height_ratios': [1, 0.4, 0.4]}, figsize=(7.5, 4.5), dpi=200)
# fig.tight_layout(pad=0.5)
fig.subplots_adjust(hspace=0.5)

ax['C'].set_xticks([])
ax['C'].set_yticks([])

for i, x in zip('DEFGHI', range(1, 7)):
  ax[i].set_xlim(-f_w, f_w)
  ax[i].set_ylim(-f_w, f_w)
  ax[i].set_xticks([])
  ax[i].set_yticks([])
  ax[i].set_aspect('equal')
  frame.append(ax[i].set_xlabel(f'$h_{x}(x)$'))

M, N = np.mgrid[-3:3:0.1, -3:3:0.1]
grid = np.dstack((M, N))
probability_grid = multivariate_normal(mean=np.array([0, 0]), cov=np.eye(2)).pdf(grid)

frame.append(ax['A'].contour(M, N, probability_grid))
frame.append(ax['A'].axhline(0, color='black', linewidth=0.2))
frame.append(ax['A'].axvline(0, color='black', linewidth=0.2))

ax['A'].set_aspect('equal')
frame.append(ax['A'].set_title('Sampling $a$'))

frame += ax['B'].plot([0, 0, 1, 1], [0, 1, 1, 0])
ax['B'].set_ylim(0, 1.5)
frame.append(ax['B'].set_title('Sampling $b$'))

frame.append(ax['C'].axhline(0, color='black', linewidth=0.2))
frame.append(ax['C'].axvline(0, color='black', linewidth=0.2))

frame.append(ax['C'].set_title(r'$a^Tx + b = c$ for $c \in \mathbb{Z}$'))

ax['C'].set_aspect('equal')
ax['C'].set_xlim(-f_w, f_w)
ax['C'].set_ylim(-f_w, f_w)

artists.append(list(frame))

for i, x in zip('DEFGHI', range(1, 7)):

  transients = []

  # Sample a
  a = rng.standard_normal(2)
  transients.append(ax['A'].scatter(*a, marker='x', color='red'))
  artists.append(list(frame + transients))

  # Sample b
  b = rng.uniform()
  transients.append(ax['B'].scatter(b, 1, marker='x', color='red'))
  artists.append(list(frame + transients))

  # Plot boundaries
  transients.append(ax['C'].arrow(0, 0, a[0], a[1], length_includes_head=True, head_width=0.05, color='black'))
  artists.append(list(frame + transients))

  p = np.array([-a[1], a[0]]) * 10 / a.min()
  lx, ly = np.array([p[0], -1*p[0]]), np.array([p[1], -1*p[1]])

  n_lines = int(f_w * np.linalg.norm(a) * np.sqrt(2)) + 1

  for c in range(-n_lines, n_lines+1):
    frame += ax['C'].plot(c*a[0]/(a@a)+lx,c*a[1]/(a@a)+ly+b, c=f'C{x}', linewidth=0.8)
    frame += ax[i].plot(c*a[0]/(a@a)+lx,c*a[1]/(a@a)+ly+b, c=f'C{x}', linewidth=0.8)

  artists.append(list(frame + transients))

frame.pop(12)
frame.pop(12)
artists.append(list(frame))

plt.close()

anim = animation.ArtistAnimation(fig, artists, interval=600, repeat=False, blit=False);
HTML(anim.to_jshtml())
```
